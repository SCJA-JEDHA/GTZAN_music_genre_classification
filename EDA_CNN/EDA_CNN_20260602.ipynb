```python
import pandas

# Format header of the output
def header_format(str:str, l) -> str:
    return "\n" + "---" + " " + str + " " + "-"*(l - len(str) - 5)

# Format end of the output
def end_format(l:int) -> str:
    return "-"*l + "\n"

def mv_prime(ds:pandas.DataFrame, l:int=30):
    # Readable output of missing values in %
    print(header_format("Dataset MValues", l))
    print(100*ds.isnull().sum()/ds.shape[0])
    print(end_format(l))

def eda_prime(ds:pandas.DataFrame, l:int=30):
    # Dataset shape & columns
    print(header_format("Dataset Shape & Columns", l))
    print(ds.shape, "\n", ds.columns)
    print(end_format(l))
    # Readable output of dataset sample(10)
    print(header_format("Dataset Sample", l))
    display(ds.sample(10))
    print(end_format(l))
    # Readable describe("all") of the dataset
    print(header_format("Dataset Describe", l))
    display(ds.describe(include = "all"))
    print(end_format(l))
    # Call of the readable output of missing values in %
    mv_prime(ds, l)
```

```python
import pandas as pd
import plotly.express as px

ds = pd.read_csv("./gtzan-dataset-music-genre-classification/Data/features_30_sec.csv", encoding = "utf-8")

eda_prime(ds)

```

```python
import torch
import torchvision.transforms as transforms
from PIL import Image
import requests
from io import BytesIO
import matplotlib.pyplot as plt
import os
import pandas as pd
import plotly.express as px

device = torch.accelerator.current_accelerator().type if torch.accelerator.is_available() else "cpu"
print(f"Using {device} device")
```

```python
ds = pd.read_csv("./gtzan-dataset-music-genre-classification/Data/features_30_sec.csv", encoding = "utf-8")
rep_image = "/content/gtzan-dataset-music-genre-classification/Data/images_original"

c_to_drop = [c for c in ds.columns if c not in ["filename", "label"]]
ds = ds.drop(columns = c_to_drop)
ds["filename"] = [str.replace(c, ".wav", ".png").replace(".0", "0") for c in ds["filename"]]
ds["path"] = [rep_image + "/" + c + "/" + f for c, f in zip(ds["label"], ds["filename"])]

occurence = ds.sample(1)
image = Image.open(*occurence["path"])
# Show the image
plt.imshow(image)
plt.title(*occurence["label"] + " : " + occurence["filename"])
plt.axis("off")  # Hide axes
plt.show()
```
